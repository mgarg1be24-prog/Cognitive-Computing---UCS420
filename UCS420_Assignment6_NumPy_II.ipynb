## Question 1
Sensor readings and vectorization.

```python
import numpy as np

temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])

# a. Add 2°C to every reading
corrected_temperature = temperature + 2
print("a. Corrected temperatures:", corrected_temperature)

# b. Convert Celsius to Fahrenheit
fahrenheit = (9/5) * corrected_temperature + 32
print("b. Fahrenheit:", fahrenheit)

# c. Readings greater than 32°C
greater_than_32 = corrected_temperature[corrected_temperature > 32]
print("c. Readings greater than 32°C:", greater_than_32)

# d. Count readings exceeding 32°C
count = np.sum(corrected_temperature > 32)
print("d. Number of readings exceeding 32°C:", count)

# e. Explanation
print("\ne. Vectorization performs operations on the complete NumPy array without")
print("explicitly writing a for loop. Boolean indexing creates a Boolean mask")
print("and directly selects values satisfying a condition, making processing")
print("shorter and generally more efficient than explicit iteration.")
```

## Question 2
Daily steps recorded by four users over three days.

```python
steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])

# a. Total steps
print("a. Total steps:", np.sum(steps))

# b. Mean number of steps
print("b. Mean steps:", np.mean(steps))

# c. Maximum and minimum
print("c. Maximum:", np.max(steps))
print("   Minimum:", np.min(steps))

# d. Total steps for each day (axis=0)
print("d. Total steps for each day:", np.sum(steps, axis=0))

# e. Total steps for each user (axis=1)
print("e. Total steps for each user:", np.sum(steps, axis=1))

# f. Position containing maximum number of steps
position = np.unravel_index(np.argmax(steps), steps.shape)
print("f. Maximum position (row, column):", position)
print("   Maximum value:", steps[position])
```

## Question 3
NumPy arrays, slicing, copying, flattening and raveling.

```python
# a. Original array
original = np.array([1, 2, 3, 4, 5, 6])
print("a. Original:", original)

# b. Slice from index 1 to 4
subset = original[1:4]
print("b. Subset:", subset)

# c. Modify first element of subset
subset[0] = 999
print("c. After modifying subset:")
print("   Original:", original)
print("   Subset:", subset)

# d. Slice using copy()
copied = original[1:4].copy()
copied[0] = 500
print("d. After modifying copied array:")
print("   Original:", original)
print("   Copied:", copied)

# e. Create 1 to 12 and reshape to 3 x 4
matrix = np.arange(1, 13).reshape(3, 4)
print("e. 3 x 4 matrix:")
print(matrix)

# f. Required indexing and slicing
print("f. First row:", matrix[0])
print("   Last row:", matrix[-1])
print("   Second column:", matrix[:, 1])
print("   Rows 1–2 and columns 2–3:")
print(matrix[1:3, 2:4])

# g. Flatten and ravel
flat_array = matrix.flatten()
ravel_array = matrix.ravel()
print("g. flatten():", flat_array)
print("   ravel():", ravel_array)

# h. Modify ravel()
ravel_array[0] = 100
print("h. After modifying ravel()[0]:")
print("   Ravel:", ravel_array)
print("   Original matrix:")
print(matrix)

# i. Modify flatten()
flat_array[1] = 200
print("i. After modifying flatten()[1]:")
print("   Flatten:", flat_array)
print("   Original matrix:")
print(matrix)

# j. Shape, ndim, size and dtype
print("j. Shape:", matrix.shape)
print("   ndim:", matrix.ndim)
print("   size:", matrix.size)
print("   dtype:", matrix.dtype)
```

## Question 4
Cognitive-assistive system using Ordinary Least Squares.

```python
X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])

y = np.array([40, 65, 30, 85])

# a. Shape and dimensions
print("a. Shape of X:", X.shape)
print("   Dimensions of X:", X.ndim)

# b. Transpose
XT = X.T
print("b. X.T:")
print(XT)

# c. Matrix product X.T @ X
XTX = X.T @ X
print("c. X.T @ X:")
print(XTX)

# d. Matrix inverse
XTX_inv = np.linalg.inv(XTX)
print("d. Inverse of X.T @ X:")
print(XTX_inv)

# e. OLS: beta = (X.T X)^(-1) X.T y
beta = XTX_inv @ X.T @ y
print("e. OLS coefficients:")
print(beta)

# f. Coefficients
print("f. Sleep hours coefficient:", beta[0])
print("   Activity level coefficient:", beta[1])
print("   Stress level coefficient:", beta[2])

# g. Prediction for new user
new_user = np.array([5, 40, 7])
predicted_score = new_user @ beta
print("g. Predicted assistance score:", predicted_score)
```

### Explanation for Question 4(f)
The three coefficients represent the estimated contribution of the three features to the assistance score in the fitted linear model, with the other features held constant.

- First coefficient → Sleep hours
- Second coefficient → Activity level
- Third coefficient → Stress level
